### Imports


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

### Load the survey data


```python
df = pd.read_csv("survey_results_public.csv")
df.head()
```

### Select columns and rename salary


```python
df = df[["Country", "EdLevel", "YearsCodePro", "Employment", "ConvertedComp"]]
df = df.rename({"ConvertedComp": "Salary"}, axis=1)
df.head()
```

### Remove missing salaries


```python
df = df[df["Salary"].notnull()]
df.head()
```

### Inspect data types


```python
df.info()
```

### Remove missing values


```python
df = df.dropna()
df.isnull().sum()
```

### Keep full-time employees


```python
df = df[df["Employment"] == "Employed full-time"]
df = df.drop("Employment", axis=1)
df.info()
```

### Count respondents by country


```python
df["Country"].value_counts()
```

### Group countries with fewer than 400 respondents


```python
def shorten_categories(categories, cutoff):
    categorical_map = {}
    for i in range(len(categories)):
        if categories.values[i] >= cutoff:
            categorical_map[categories.index[i]] = categories.index[i]
        else:
            categorical_map[categories.index[i]] = 'Other'
    return categorical_map
country_map = shorten_categories(df.Country.value_counts(), 400)
df['Country'] = df['Country'].map(country_map)
df.Country.value_counts()
```

### Plot salaries by country


```python
fig, ax = plt.subplots(1,1, figsize=(12, 7))
df.boxplot('Salary', 'Country', ax=ax)
plt.suptitle('Salary (US$) v Country')
plt.title('')
plt.ylabel('Salary')
plt.xticks(rotation=90)
plt.show()
```

### Filter salary outliers and plot again


```python
df = df[df["Salary"] <= 250000]
df = df[df["Salary"] >= 10000]
df = df[df['Country'] != 'Other']
fig, ax = plt.subplots(1,1, figsize=(12, 7))
df.boxplot('Salary', 'Country', ax=ax)
plt.suptitle('Salary (US$) v Country')
plt.title('')
plt.ylabel('Salary')
plt.xticks(rotation=90)
plt.show()
```

### Inspect professional experience values


```python
df["YearsCodePro"].unique()
```

### Convert experience to numbers


```python
def clean_experience(x):
    if x ==  'More than 50 years':
        return 50
    if x == 'Less than 1 year':
        return 0.5
    return float(x)

df['YearsCodePro'] = df['YearsCodePro'].apply(clean_experience)
```

### Inspect education values


```python
df["EdLevel"].unique()
```

### Group education levels


```python
def clean_education(value):
    if "Bachelor" in value:
        return "Bachelor's degree"
    if "Master" in value:
        return "Master's degree"
    if "Professional degree" in value or "Other doctoral" in value:
        return "Post grad"
    return "Less than a Bachelors"

df["EdLevel"] = df["EdLevel"].apply(clean_education)
df["EdLevel"].unique()
```

### Encode education


```python
from sklearn.preprocessing import LabelEncoder
le_education = LabelEncoder()
df['EdLevel'] = le_education.fit_transform(df['EdLevel'])
df["EdLevel"].unique()
```

### Encode countries


```python
le_country = LabelEncoder()
df['Country'] = le_country.fit_transform(df['Country'])
df["Country"].unique()
```

### Prepare features and target


```python
X = df.drop("Salary", axis=1)
y = df["Salary"]
```

### Train linear regression


```python
from sklearn.linear_model import LinearRegression
linear_reg = LinearRegression()
linear_reg.fit(X, y.values)
```

### Evaluate linear regression (training RMSE)


```python
y_pred = linear_reg.predict(X)
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
error = np.sqrt(mean_squared_error(y, y_pred))

error
```

### Train a decision tree


```python
from sklearn.tree import DecisionTreeRegressor
dec_tree_reg = DecisionTreeRegressor(random_state=0)
dec_tree_reg.fit(X, y.values)
```

### Evaluate the decision tree (training RMSE)


```python
y_pred = dec_tree_reg.predict(X)
error = np.sqrt(mean_squared_error(y, y_pred))
print("${:,.02f}".format(error))
```

### Train a random forest


```python
from sklearn.ensemble import RandomForestRegressor
random_forest_reg = RandomForestRegressor(random_state=0)
random_forest_reg.fit(X, y.values)
```

### Evaluate the random forest (training RMSE)


```python
y_pred = random_forest_reg.predict(X)
error = np.sqrt(mean_squared_error(y, y_pred))
print("${:,.02f}".format(error))
```

### Tune the decision tree with cross-validation


```python
from sklearn.model_selection import GridSearchCV

max_depth = [None, 2,4,6,8,10,12]
parameters = {"max_depth": max_depth}

regressor = DecisionTreeRegressor(random_state=0)
gs = GridSearchCV(regressor, parameters, scoring='neg_mean_squared_error')
gs.fit(X, y.values)
```

### Evaluate the best tree (training RMSE)


```python
regressor = gs.best_estimator_

regressor.fit(X, y.values)
y_pred = regressor.predict(X)
error = np.sqrt(mean_squared_error(y, y_pred))
print("${:,.02f}".format(error))
```

### Predict an example salary


```python
sample = pd.DataFrame({
    "Country": le_country.transform(["United States"]),
    "EdLevel": le_education.transform(["Master's degree"]),
    "YearsCodePro": [15.0],
})
y_pred = regressor.predict(sample)
y_pred
```

```python
import pickle
data = {"model": regressor, "le_country": le_country, "le_education": le_education}
with open('saved_steps.pkl', 'wb') as file:
    pickle.dump(data, file)
```
